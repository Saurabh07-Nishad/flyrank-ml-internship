## Target / proxy

The provisional target is a content opportunity proxy: a page that shows meaningful search demand but weak click-through performance may deserve review. For example, pages with at least 100 impressions and very low or zero clicks can be treated as candidate opportunities. This is a defined rule-based proxy from observable data, not a ground-truth measure of whether a refresh will succeed. Later validation would be needed to test whether higher-scored pages actually produce better outcomes after an action.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

data_path = "../../data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

print("Number of pages:", len(df))
print("Number of columns:", len(df.columns))

opportunity_pages = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print("Pages with 100+ impressions but 0 clicks:",
      len(opportunity_pages))
```

## Success metric

The main success metric will be Precision@K for the ranked review queue. It asks how many of the top K pages selected by the scoring method are genuinely useful review opportunities according to the chosen evaluation outcome. A higher Precision@K means the limited review effort is being concentrated on more relevant pages. For the initial framing, the score is decision-support and directional rather than proof that a refresh will succeed.

## Why ML beats a fixed rule

A fixed rule such as "100+ impressions and 0 clicks" is easy to understand, but it treats every page that meets the threshold in the same way. In reality, opportunity can depend on several signals together, including impressions, CTR, average position, content age, freshness, sessions, and recent trends. ML can learn combinations of these signals and produce a more flexible ranking than a small set of manually chosen if-statements. The fixed rule is still useful as a simple baseline that a later model should have to beat.
