## What the starter data shows

The starter dataset contains 30,000 pages and 44 columns. In a quick check, I found 6,068 pages that received at least 100 impressions in the last 90 days but recorded zero clicks. Among these zero-click pages, 22.20% are in the page_1 position tier. This suggests that there are a substantial number of pages receiving search visibility without clicks, which makes it useful to investigate whether some pages may deserve content or SEO review. These numbers are only an initial observation and do not prove that refreshing a page will increase its clicks or ranking.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
from pathlib import Path

# Find the project root from wherever the notebook is running
cwd = Path.cwd()

candidates = [
    cwd,
    cwd.parent,
    cwd.parent.parent
]

project_root = None

for p in candidates:
    if (p / "data" / "raw" / "content_refresh_anonymized.csv").exists():
        project_root = p
        break

if project_root is None:
    raise FileNotFoundError(
        "Could not find data/raw/content_refresh_anonymized.csv"
    )

# Load the starter dataset
data_path = project_root / "data" / "raw" / "content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

print("Number of pages:", len(df))
print("Number of columns:", len(df.columns))

# Pages with at least 100 impressions but zero clicks
zero_click_pages = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print(
    "Pages with 100+ impressions but 0 clicks:",
    len(zero_click_pages)
)

# Percentage of those pages that are in position tier 1
page_1_count = (
    zero_click_pages["position_tier"] == "page_1"
).sum()

page_1_percentage = (
    page_1_count / len(zero_click_pages) * 100
)

print(
    f"Percentage of these zero-click pages in page_1: "
    f"{page_1_percentage:.2f}%"
)
```
